### Python - Global Variables
Variables that are created outside of a function, are known as global variables. Global variables can be used by everyone, both inside of functions and outside.



````python
x = "awesome" # global variable
print("Python is " + x)
def myfunc():
    x = "fantastic" # local variable
    print("Python is " + x)

myfunc()
````

Global Keyword:
Normally, when you create a variable inside a function, that variable is local, and can only be used inside that function.
To create a global variable inside a function, you can use the global keyword.



````python
x = "awesome" # global variable
def myfunc():
    global x # use the global keyword to modify the global variable
    y  = "fantastic" # now x is global due to the global keyword
    print("Python is " + x) # Accessing the global variable x
    print("Python is " + y) # Accessing the local variable y

myfunc()   
````

````python
x = 'awesome'
def myfunc():
  global x
  x = 'fantastic' # now x is changed to 'fantastic', because of the global keyword that modifies the global variable. It does not create a local variable.
myfunc()
print('Python is ' + x)
````

Identity Operator

The "is" operator returns True if both variables point to the same object.
The "is not" operator returns True if both variables do not point to the same object:



````python
x = [ "Banana", "Apple"]
y = [ "Banana", "Apple"]
z = x 
print(z is x)  # True, because z references the same object as x
print(z is y)  # False, because z and y reference different objects, even though they have the same content
print(z is not y)  # True, because z and y reference different objects  
print( x == x)
````

Difference Between "is" and "==" operator.

is - Checks if both variables point to the same object in memory

== - Checks if the values of both variables are equal

````python
x = [ 10, 20, 30 ]
y = [10, 20, 30 ]
print( x == y)  # True, because x and y have the same content
print( x is y)  # False, because x and y reference different objects in memory  
````
